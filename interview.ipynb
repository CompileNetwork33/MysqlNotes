```python
The denominator ignores the Product filter while retaining other relevant filters.”

4️⃣ What is the difference between calculated columns, measures, and calculated tables?

Sample Answer:

“A calculated column creates a value for every row and is stored in the model.

A measure is calculated dynamically based on the current filter context and is generally preferred for aggregations and KPIs.

A calculated table creates a new table using DAX.

For example:

Calculated column → Profit for each transaction

Measure → Total Profit

Calculated table → A DAX-generated table used for a specific modeling requirement.”

5️⃣ What is cardinality in Power BI relationships?

Sample Answer:

“Cardinality describes how records in one table relate to records in another table.

Common relationship types include:

• One-to-many

• Many-to-one

• One-to-one

• Many-to-many  

In a typical star schema, a dimension table has unique keys and the fact table contains repeated keys, resulting in a one-to-many relationship from the dimension to the fact table.”

6️⃣ What is the difference between a single-direction and bidirectional relationship?

Sample Answer:

“In a single-direction relationship, filters generally flow from one table to another in the defined direction.

In a bidirectional relationship, filters can propagate in both directions.

I generally prefer single-direction relationships where possible because they make the model easier to understand and can reduce ambiguity and unexpected filtering behavior.

Bidirectional relationships should be used when there is a clear modeling requirement.”

7️⃣ What is a Date table and why is it important?

Sample Answer:

“A Date table is a dedicated table containing dates and related attributes such as year, quarter, month, week, and day.

It is important because consistent time-based analysis and time-intelligence calculations depend on a proper date structure.

For example, instead of repeatedly extracting the month and year from a transaction date, I can use fields from a dedicated Date dimension.”

8️⃣ What is the difference between Import mode and DirectQuery?

Sample Answer:

“In Import mode, data is loaded into Power BI's in-memory model. This generally provides fast report performance, but the data needs to be refreshed.

In DirectQuery, Power BI sends queries to the underlying data source when users interact with the report. This can allow access to more current data without importing the entire dataset, but performance depends heavily on the source system and query design.

The choice depends on factors such as data volume, freshness requirements, source performance, and modeling limitations.”

9️⃣ What is a many-to-many relationship, and why can it be problematic?

Sample Answer:

“A many-to-many relationship occurs when multiple records

 in one table can relate to multiple records in another table.

It can be useful in certain business scenarios, but it can also introduce ambiguous filter paths and unexpected results if the model isn't designed carefully.

Where possible, I would look for a cleaner model, such as using a bridge table, to represent the many-to-many relationship explicitly.”

🔟 How would you calculate the percentage of total sales in Power BI?

Sample Answer:

“I would create a total sales measure and then calculate the denominator after removing the appropriate product or category filter.”

Total Sales =
SUM(Sales[Sales Amount])

Sales % of Total =
DIVIDE(
    [Total Sales],
    CALCULATE(
        [Total Sales],
        REMOVEFILTERS(Product)
    )
)

“This allows the numerator to respond to the current product context while the denominator represents the overall total across products.”
```

```python
📊 Data Analyst Interview Series — Part 13

📊 Power BI Interview Questions

Part 13 — Power BI Fundamentals

1️⃣ What is Power BI and why is it used?

Sample Answer:

Power BI is a business intelligence and data visualization tool from Microsoft. It helps analysts connect to different data sources, clean and transform data, create data models, build interactive reports and dashboards, and share insights with stakeholders.

For example, I can connect Power BI to SQL Server, Excel, or CSV files, transform the data using Power Query, create a data model, write DAX measures, and build an interactive sales dashboard.

2️⃣ What are the main components of Power BI?

Sample Answer:

The major components I commonly work with are:

• Power BI Desktop — Build reports and data models.
• Power BI Service — Publish, share, and manage reports online.
• Power BI Mobile — View reports on mobile devices.
• Power Query — Connect, clean, and transform data.
• DAX — Create calculations and measures.
• On-premises Data Gateway — Connect Power BI Service to on-premises data sources.

Together, these components support the complete reporting and analytics process.

3️⃣ What is the difference between Power BI Desktop and Power BI Service?

Sample Answer:

Power BI Desktop is mainly used for developing reports and data models, while Power BI Service is a cloud-based platform used to publish, share, collaborate, and manage those reports.

For example:

Power BI Desktop → Build the report → Publish → Power BI Service → Share with users.

4️⃣ What is Power Query in Power BI?

Sample Answer:

Power Query is used for extracting, cleaning, and transforming data before loading it into the Power BI model.

For example, I can use Power Query to:

• Remove duplicates
• Handle missing values
• Change data types
• Split columns
• Merge datasets
• Append datasets
• Filter records
• Create calculated transformation columns

Power Query is essentially used for the ETL part of the process.

5️⃣ What is the difference between Power Query and DAX?

Sample Answer:

Power Query is primarily used to prepare and transform data before it enters the model, whereas DAX is used to create calculations within the Power BI model.

For example:

Power Query → Remove duplicates, merge tables, clean customer names.

DAX → Calculate Total Sales, Profit Margin, Year-over-Year Growth.

A simple way to remember it is:

Power Query = Data preparation

DAX = Data analysis and calculations

6️⃣ What is a data model in Power BI?

Sample Answer:

A data model defines how different tables are structured and connected within Power BI.

For example, I might have:

Sales  
Customers  
Products  
Date

The Sales table can be connected to Customers, Products, and Date using relationships.

A well-designed data model makes reporting more accurate, easier to maintain, and more efficient.

7️⃣ What is a relationship in Power BI?

Sample Answer:

A relationship connects two tables using a common column.

For example:

Customers  
↓  
Sales[CustomerID]

This allows Power BI to understand which sales transactions belong to which customers.

Common relationship types include:

• One-to-many
• Many-to-one
• One-to-one
• Many-to-many

In most analytical models, one-to-many relationships are commonly used.

8️⃣ What is a star schema and why is it important?

Sample Answer:

A star schema is a data modeling approach where a central fact table is connected to multiple dimension tables.

For example:

FactSales  
→ DimCustomer  
→ DimProduct  
→ DimDate  
→ DimRegion

The fact table contains measurable business transactions, while dimension tables provide descriptive information.

I prefer a star schema because it makes the model easier to understand, improves filtering behavior, and generally supports better reporting performance.
```
