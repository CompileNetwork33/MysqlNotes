```python
The denominator ignores the Product filter while retaining other relevant filters.”

4️⃣ What is the difference between calculated columns, measures, and calculated tables?

Sample Answer:

“A calculated column creates a value for every row and is stored in the model.

A measure is calculated dynamically based on the current filter context and is generally preferred for aggregations and KPIs.

A calculated table creates a new table using DAX.

For example:

Calculated column → Profit for each transaction

Measure → Total Profit

Calculated table → A DAX-generated table used for a specific modeling requirement.”

5️⃣ What is cardinality in Power BI relationships?

Sample Answer:

“Cardinality describes how records in one table relate to records in another table.

Common relationship types include:

• One-to-many

• Many-to-one

• One-to-one

• Many-to-many  

In a typical star schema, a dimension table has unique keys and the fact table contains repeated keys, resulting in a one-to-many relationship from the dimension to the fact table.”

6️⃣ What is the difference between a single-direction and bidirectional relationship?

Sample Answer:

“In a single-direction relationship, filters generally flow from one table to another in the defined direction.

In a bidirectional relationship, filters can propagate in both directions.

I generally prefer single-direction relationships where possible because they make the model easier to understand and can reduce ambiguity and unexpected filtering behavior.

Bidirectional relationships should be used when there is a clear modeling requirement.”

7️⃣ What is a Date table and why is it important?

Sample Answer:

“A Date table is a dedicated table containing dates and related attributes such as year, quarter, month, week, and day.

It is important because consistent time-based analysis and time-intelligence calculations depend on a proper date structure.

For example, instead of repeatedly extracting the month and year from a transaction date, I can use fields from a dedicated Date dimension.”

8️⃣ What is the difference between Import mode and DirectQuery?

Sample Answer:

“In Import mode, data is loaded into Power BI's in-memory model. This generally provides fast report performance, but the data needs to be refreshed.

In DirectQuery, Power BI sends queries to the underlying data source when users interact with the report. This can allow access to more current data without importing the entire dataset, but performance depends heavily on the source system and query design.

The choice depends on factors such as data volume, freshness requirements, source performance, and modeling limitations.”

9️⃣ What is a many-to-many relationship, and why can it be problematic?

Sample Answer:

“A many-to-many relationship occurs when multiple records

 in one table can relate to multiple records in another table.

It can be useful in certain business scenarios, but it can also introduce ambiguous filter paths and unexpected results if the model isn't designed carefully.

Where possible, I would look for a cleaner model, such as using a bridge table, to represent the many-to-many relationship explicitly.”

🔟 How would you calculate the percentage of total sales in Power BI?

Sample Answer:

“I would create a total sales measure and then calculate the denominator after removing the appropriate product or category filter.”

Total Sales =
SUM(Sales[Sales Amount])

Sales % of Total =
DIVIDE(
    [Total Sales],
    CALCULATE(
        [Total Sales],
        REMOVEFILTERS(Product)
    )
)

“This allows the numerator to respond to the current product context while the denominator represents the overall total across products.”
```
