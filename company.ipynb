SQL AGGREGATE FUNCTIONS

👉 Aggregate function multiple rows par kaam karke single result deta hai.

🔹 Common Aggregate Functions

1️⃣ COUNT()
2️⃣ SUM()
3️⃣ AVG()
4️⃣ MIN()
5️⃣ MAX()


```python
cursor.execute("SELECT COUNT(*) FROM employees")
print(cursor.fetchone())
```

```python
cursor.execute("SELECT SUM(salary) FROM employees")
print(cursor.fetchone())
```

```python
cursor.execute("SELECT AVG(salary) FROM employees")
print(cursor.fetchone())
```

cursor.execute("""
SELECT department, SUM(salary)
FROM employees
GROUP BY department
""")

rows = cursor.fetchall()
for row in rows:
    print(row)


```python
cursor.execute("""
SELECT department 
               FROM employees
               GROUP BY department
""")

rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
cursor.execute("""
SELECT department, COUNT(age) FROM employees GROUP BY department  HAVING age>32
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

HAVING CLAUSE (Easy words)

👉 HAVING = GROUP BY ke baad condition lagana
👉 Jab aggregate function par condition chahiye → HAVING

❌ WHERE vs HAVING (sabse important)
WHERE	HAVING
Rows filter karta	Groups filter karta
Aggregate use ❌	Aggregate use ✅
GROUP BY se pehle	GROUP BY ke baad

```python
cursor.execute("""
SELECT e.name, d.dept_name, e.salary
FROM employees e
JOIN departments d
ON e.dept_id = d.dept_id
""")

for row in cursor.fetchall():
    print(row)
```
