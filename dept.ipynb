```python
import sqlite3
conn = sqlite3.connect("department.db")
cursor= conn.cursor()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS department(
              dept_id INT PRIMARY KEY,
               dept_name VARCHAR,
               location VARCHAR,
               director VARCHAR
                 )""")
conn.commit()
print("Table created successfully")
```

```python
cursor.execute('''
INSERT INTO department(dept_id,dept_name,location,director)
               VALUES(1,'hr','Lucknow','Vaishnavi'),
               (3,'Civil','Delhi','Manas'),
               (5,'Mechanical','Mumbai','Riya'),
               (4,'Electrical','Chennai','Arav'),
               (7,'CSE','Bangalore','Kamal'),
               (8,'Cyber',"Pune",'Nisha')''')
conn.commit()
```

```python
cursor.execute("SELECT*FROM department")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
cursor.execute("SELECT* FROM department WHERE dept_name IN ('CSE','hr')")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
cursor.execute("SELECT* FROM department WHERE Director LIKE 'M%'")
rows = cursor.fetchall()
for row in rows:
    print(row)#name starts with m
```

```python
#name ends with a
cursor.execute("SELECT* FROM department WHERE director LIKE'%a'")
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees(
               id INTEGER PRIMARY KEY AUTOINCREMENT,
               name TEXT,
               salary INTEGER,
               department TEXT,
               age INTEGER)

               """)
conn.commit()
```

```python
cursor.execute("""
SELECT e.name, d.department, e.salary
FROM employees e
JOIN department d
ON e.dept_name = d.dept_name
""")
rows = cursor.fetchall()
for row in rows:
    print(row)
```
