```python
🐍 Data Analyst Interview Series: Part 31

Guys, let's continue our Python Interview Series.

In the previous part, we covered Pandas GroupBy and Aggregation. Today, let's focus on combining datasets and reshaping data using Pandas.

Part 31, Pandas Merge, Join, Concat & Pivot Tables

1️⃣ What is the difference between merge() and concat() in Pandas?

Sample Answer:

Both functions combine DataFrames, but they work differently.

merge() combines DataFrames based on common columns or indexes, similar to SQL JOIN operations.

Example:

import pandas as pd

df1 = pd.DataFrame({
    "CustomerID": [1, 2, 3],
    "Name": ["Rahul", "Priya", "Amit"]
})

df2 = pd.DataFrame({
    "CustomerID": [1, 2, 3],
    "Sales": [5000, 7000, 6000]
})

result = pd.merge(df1, df2, on="CustomerID")

concat() combines DataFrames along rows or columns.

result = pd.concat([df1, df2], axis=0)

In simple terms:

Merge → Combine based on matching keys.

Concat → Stack DataFrames along rows or columns.

2️⃣ What are the different types of joins in Pandas?

Pandas supports different types of joins through the merge() function.

• Inner Join, Returns matching records from both DataFrames.

• Left Join, Returns all records from the left DataFrame and matching records from the right.

• Right Join, Returns all records from the right DataFrame and matching records from the left.

• Outer Join, Returns all records from both DataFrames.

result = pd.merge(
    df1,
    df2,
    on="CustomerID",
    how="left"
)

This keeps all customers from df1, even if they don't have matching records in df2.

These joins are useful when combining customer, sales, product, and transaction datasets.

3️⃣ What is the difference between join() and merge()?

merge() is primarily used to combine DataFrames based on columns or indexes.

join() is primarily used to combine DataFrames using their indexes by default.

result = df1.join(
    df2,
    how="left"
)

We can also specify columns for joining using join().

I generally use merge() when joining datasets using common business keys such as CustomerID or ProductID.

4️⃣ What happens when both DataFrames contain columns with the same name?

When merging DataFrames with overlapping non-key column names, Pandas automatically adds suffixes.

result = pd.merge(
    df1,
    df2,
    on="CustomerID",
    suffixes=("_left", "_right")
)

This helps distinguish columns coming from different DataFrames.

It is useful when combining datasets that contain common fields such as Status, Date, or Amount.

5️⃣ What is a Pivot Table in Pandas?

A Pivot Table is used to summarize and reorganize data based on different categories.

It is similar to PivotTables in Excel.

df = pd.DataFrame({
    "Region": ["North", "South", "North", "South"],
    "Product": ["Laptop", "Laptop", "Mobile", "Mobile"],
    "Sales": [50000, 60000, 30000, 40000]
})

pivot = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Product",
    aggfunc="sum",
    fill_value=0
)

print(pivot)

This creates a summary of sales by region and product.

6️⃣ What is the difference between pivot() and pivot_table()?

pivot() reshapes data but does not perform aggregation. It requires each index-column combination to be unique.

pivot_table() can handle duplicate combinations by applying an aggregation function such as sum, mean, or count.

df.pivot(
    index="Region",
    columns="Product",
    values="Sales"
)

Using pivot_table():

df.pivot_table(
    index="Region",
    columns="Product",
    values="Sales",
    aggfunc="sum"
)
```

```python
🐍 Data Analyst Interview Series — Part 32 Advanced Pandas Operations

Today, we'll cover some important advanced Pandas concepts frequently asked in Data Analyst interviews.

1️⃣ What is the difference between apply(), map(), and DataFrame.map()?

Sample Answer:

- map() is used on a Series to map or replace individual values.
- apply() applies a function to a Series or along rows or columns of a DataFrame.
- DataFrame.map() applies a function element-wise to every value in a DataFrame.

Example:

df["salary"].map(lambda x: x * 1.10)
2️⃣ What is the difference between apply() and transform() in GroupBy?

Sample Answer:

- apply() offers flexibility and can return different shapes of output.
- transform() returns results aligned with the original rows, making it useful for creating new columns.

Example:

df["dept_avg"] = (
    df.groupby("department")["salary"]
      .transform("mean")
)
This adds the average salary of each department to every employee's row.

3️⃣ What is rolling() in Pandas?

Sample Answer:

rolling() performs calculations over a moving window of observations.

It is commonly used to calculate moving averages and smooth time-series data.

Example:

df["moving_avg"] = (
    df["sales"].rolling(window=3).mean()
)
This calculates the average sales of the current row and previous two rows.

4️⃣ What is the difference between rolling() and expanding()?

Sample Answer:

- rolling() calculates values over a fixed-size window.
- expanding() calculates values using all observations from the beginning up to the current row.

Example:

df["rolling_avg"] = df["sales"].rolling(3).mean()

df["expanding_avg"] = df["sales"].expanding().mean()
5️⃣ How do you calculate cumulative sales in Pandas?

Sample Answer:

We use cumsum() to calculate the running total.

Example:

df["cumulative_sales"] = df["sales"].cumsum()
For department-wise cumulative sales:

df["cumulative_sales"] = (
    df.groupby("department")["sales"]
      .cumsum()
)
6️⃣ How do you calculate percentile rank in Pandas?

Sample Answer:

We use rank(pct=True) to calculate the relative rank of each value.

Example:

df["sales_percentile"] = (
    df["sales"].rank(pct=True) * 100
)
This assigns each sales value a percentile rank.

7️⃣ What is shift() in Pandas?

Sample Answer:

shift() moves values up or down by a specified number of rows.

It is useful for comparing current values with previous values.

Example:

df["previous_sales"] = df["sales"].shift(1)

df["sales_change"] = (
    df["sales"] - df["previous_sales"]
)
8️⃣ What is resample() in Pandas?

Sample Answer:

resample() is used to group time-series data into different time frequencies, such as daily, weekly, or monthly.

Example:

df["date"] = pd.to_datetime(df["date"])

df = df.set_index("date")

monthly_sales = df["sales"].resample("MS").sum()``
Here, MS groups sales by month, starting at the beginning of each month.

9️⃣ How do you calculate month-over-month sales growth?

Sample Answer:

First, aggregate sales by month. Then use pct_change() to calculate the percentage change from the previous month.

Example:

df["date"] = pd.to_datetime(df["date"])

monthly_sales = (
    df.set_index("date")["sales"]
      .resample("MS")
      .sum()
)

monthly_sales_growth = monthly_sales.pct_change() * 100
This calculates the month-over-month percentage growth in sales.

🔟 How do you calculate the difference between the current and previous month's sales for each product?

Sample Answer:

We can use groupby() with shift().

Example:
df = df.sort_values(["product", "month"])

df["previous_sales"] = (
    df.groupby("product")["sales"].shift(1)
)

df["sales_difference"] = (
    df["sales"] - df["previous_sales"]
)
This compares each product's sales with its previous recorded month.
```
