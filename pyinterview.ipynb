```python
🐍 Data Analyst Interview Series — Part 23

Guys, let's start the Python section of our Data Analyst Interview Series.

We'll begin with Python fundamentals and gradually move towards Pandas, NumPy, data cleaning, data analysis, and practical interview questions.

Part 23 — Python Fundamentals

1️⃣ What is Python and why is it used in Data Analytics?

Sample Answer:

Python is a high-level, interpreted programming language known for its simple syntax and extensive libraries.

In Data Analytics, Python is used for:

• Data cleaning and transformation

• Data manipulation

• Exploratory Data Analysis (EDA)

• Data visualization

• Statistical analysis

• Machine learning

• Automation  

Libraries like Pandas, NumPy, Matplotlib, and Seaborn make Python especially useful for data analysis.

2️⃣ What are the main data types in Python?

Sample Answer:

Python has several built-in data types.

The most commonly used are:

• int — Integer values

• float — Decimal values

• str — Text values

• bool — True or False

• list — Ordered, mutable collection

• tuple — Ordered, immutable collection

• set — Collection of unique values

• dict — Key-value pairs  

Example:

age = 25
salary = 45000.50
name = "Rahul"
is_employed = True

I use these data types to store and manipulate different kinds of information.

3️⃣ What is the difference between a list and a tuple?

Sample Answer:

Both lists and tuples are used to store multiple values.

List:

• Mutable

• Uses square brackets []

• Elements can be modified.

Tuple:

• Immutable

• Uses parentheses ()

• Elements cannot be modified after creation.

Example:

my_list = [10, 20, 30]
my_tuple = (10, 20, 30)

I use lists when data needs modification and tuples when I want to maintain a fixed collection.

4️⃣ What is the difference between a set and a list?

Sample Answer:

A list maintains order and allows duplicate values.

A set stores unique elements and does not guarantee a particular iteration order.

Example:

numbers = [1, 2, 2, 3, 3, 4]

unique_numbers = set(numbers)

Output:

{1, 2, 3, 4}

Sets are useful when I need to remove duplicates or perform operations like union and intersection.

5️⃣ What is a dictionary in Python?

Sample Answer:

A dictionary stores data in key-value pairs.

Example:

employee = {
    "name": "Rahul",
    "age": 28,
    "department": "Analytics"
}

print(employee["name"])

Output:

Rahul

Dictionaries are useful for storing structured information and retrieving values using keys.

6️⃣ What is the difference between == and is in Python?

Sample Answer:

The == operator checks whether two objects have equal values.

The is operator checks whether two references point to the same object.

Example:

a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)  # True
print(a is b)  # False

Both lists contain the same values, but they are different objects in memory.

7️⃣ What is the difference between break, continue, and pass?

Sample Answer:

These are control-flow statements used in Python.

• break — Terminates the loop.

• continue — Skips the current iteration and moves to the next one.

• pass — Does nothing; it acts as a placeholder.  

Example:

for i in range(5):
    if i == 2:
        continue
    print(i)

Output:

0
1
3
4
```

```python
8️⃣ What is the difference between a for loop and a while loop?

Sample Answer:

A for loop is generally used to iterate over a sequence or iterable.

A while loop continues executing as long as a condition is True.

Example:

For loop:

for i in range(5):
    print(i)

While loop:

i = 0

while i < 5:
    print(i)
    i += 1

I use for loops when iterating over collections and while loops when execution depends on a condition.

9️⃣ What is a function in Python?

Sample Answer:

A function is a reusable block of code designed to perform a specific task.

It helps reduce code duplication and makes programs easier to maintain.

Example:

def calculate_total(a, b):
    return a + b

result = calculate_total(10, 20)

print(result)

Output:

30

Functions are useful for organizing repetitive data-processing tasks.

🔟 What is the difference between local and global variables?

Sample Answer:

A local variable is defined inside a function and is generally accessible only within that function.

A global variable is defined outside functions and can be accessed throughout the module.

Example:

x = 100

def my_function():
    y = 50
    print(x)
    print(y)

my_function()

Here, x is a global variable and y is a local variable.

I generally prefer local variables where possible because they make code easier to understand and maintain.
```

```python
🐍 Data Analyst Interview Series — Part 24

Guys, let's continue our Python Interview Series.

In this part, we'll cover Python strings, list comprehensions, lambda functions, and commonly asked coding questions.

Part 24 — Python Core Concepts

1️⃣ What is string slicing in Python?

Sample Answer:

String slicing allows us to extract a portion of a string using index positions.

Syntax:

string[start:stop:step]

Example:

text = "Python Programming"

print(text[0:6])
print(text[::-1])

Output:

Python
gnimmargorP nohtyP

It is useful when extracting, modifying, or reversing text data.

2️⃣ What is the difference between split() and join()?

Sample Answer:

split() breaks a string into a list based on a separator.

join() combines multiple strings into a single string using a separator.

Example:

text = "Python SQL Excel"

words = text.split(" ")
print(words)

Output:

['Python', 'SQL', 'Excel']

Using join():

result = "-".join(words)
print(result)

Output:

Python-SQL-Excel

These methods are useful for text cleaning and formatting.

3️⃣ What is list comprehension in Python?

Sample Answer:

List comprehension provides a concise way to create lists using an expression and an iterable.

Example:

numbers = [1, 2, 3, 4, 5]

squares = [x**2 for x in numbers]

print(squares)

Output:

[1, 4, 9, 16, 25]

It can also include conditions:

even_numbers = [x for x in numbers if x % 2 == 0]

Output:

[2, 4]

List comprehensions are useful for simple data transformations.

4️⃣ What is a lambda function?

Sample Answer:

A lambda function is a small anonymous function written in a single expression.

Example:

square = lambda x: x * x

print(square(5))

Output:

25

Lambda functions are commonly used when a short function is needed temporarily.

For example:

employees = [
    ("Rahul", 50000),
    ("Amit", 65000),
    ("Priya", 60000)
]

employees.sort(key=lambda x: x[1])

This sorts employees by salary in ascending order.

5️⃣ What is the difference between map(), filter(), and reduce()?

Sample Answer:

These functions are used for processing iterable data.

map() applies a function to every element.

numbers = [1, 2, 3, 4]

result = list(map(lambda x: x * 2, numbers))

Output:

[2, 4, 6, 8]

filter() selects elements that satisfy a condition.

result = list(filter(lambda x: x > 2, numbers))

Output:

[3, 4]

reduce() repeatedly combines elements into a single result.

from functools import reduce

result = reduce(lambda x, y: x + y, numbers)

Output:

10

6️⃣ What is the difference between append() and extend()?

Sample Answer:

append() adds one element to the end of a list.

extend() adds all elements from another iterable.

Example:

a = [1, 2]

a.append([3, 4])
print(a)

Output:

[1, 2, [3, 4]]

Using extend():

b = [1, 2]

b.extend([3, 4])
print(b)

Output:

[1, 2, 3, 4]
```
